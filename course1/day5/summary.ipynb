## (3) 自分の写真ならどうするか

演習2（`ex2.ipynb`）の結果を写す．

|---|---|---|
| CNN（ゼロから） | 63.3 | ― |

- ResNet18 なら，何層まで学習し直すのがよいと考えるか．理由：写真が少ないため最終層のみで学習し直す
